## 4. Data preparation

```python
x = diabets_data.drop("Outcome",axis=1)
y = diabets_data["Outcome"]
```

## 5. Model Building

```python
model = Sequential()
```

```python
model.add(Dense(units=100,input_dim=8,kernel_initializer='uniform',activation="relu"))
model.add(Dense(units=68,kernel_initializer='uniform',activation="relu"))
model.add(Dense(1,kernel_initializer='uniform',activation="sigmoid"))
```

```python
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,stratify=y)
```

```python
x_train.shape,x_test.shape
```

```python
std = StandardScaler()
x_train = std.fit_transform(x_train)
x_test = std.transform(x_test)
```

```python
model.fit(x=x_train,y=y_train,batch_size=16,epochs=50,validation_split=0.2)
```

```python
y_pred = model.predict(x_test)
y_pred = (y_pred>0.6).astype(int)
y_pred
```

```python
model.save("diabetes.h5")
```
